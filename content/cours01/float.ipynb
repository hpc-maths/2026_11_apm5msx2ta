# Représentation des réels en machine 

Le module Python mpmath permet de manipuler des flottants à precision arbitraire en fixant la taille de la mantisse. La documentation de ce module est disponible [ici](https://mpmath.org/doc/current/).

```python
a = mp.mpf(1.5)
print(a)
```

## Perte de l'associativité

$(a - b ) + c = (a + c) - b$ ?

```python
# simple precision prec=24
# double precision prec=53
mp.prec = 24

a = mp.mpf('1000000')
b = mp.mpf('999999.5')
c = mp.mpf('0.1')

res1 = (a - b) + c
res2 = (a + c) - b

print(f"(a - b) + c = {res1}")
print(f"(a + c) - b = {res2}")
```

## Calcul d'une somme 

On calcule la somme :  $\displaystyle 1  + \frac{1}{2} + \frac{1}{3} + \dots + \frac{1}{10^6}$ d'abord dans cet ordre puis dans l'ordre inverse :

```python
# simple precision prec=24
# double precision prec=53
mp.prec = 24

s1 = mp.mpf('0')
for i in range(1,1_000_001): s1 += mp.mpf(1./i)
print(f"s1 = {s1}")

s2 = mp.mpf('0')
for i in range(1_000_000, 0, -1): s2 += mp.mpf(1./i)
print(f"s2 = {s2}")

print(f"|s1-s2| = {(abs(s1-s2)/s2)*100} %")
```

## Evaluation d'une fonction 

On considère l'évaluation de la fonction $g(x)$ écrite sous deux formes différentes :

$$g_1(x) = \displaystyle \frac{1}{x(x+1)} \quad \text{et} \quad g_2(x) = \displaystyle \frac{1}{x} - \frac{1}{x+1}$$ 

```python
def g1(x):
    return 1/(x*(x+1))

def g2(x):
    return 1/x - 1/(x+1)
```

**Evaluation de la fonction $g(x)$ pour $x = 100 0000 000$**

```python
# Evaluation de la valeur de référence en utilisant des flottants quadruple précision
mp.prec = 113
xref = mp.mpf('100_000_000')
gref = g1(xref)
print(f"Valeur de réference avec {mp.dps} chiffres significatifs = {gref}")

# simple precision
mp.prec = 24
print(f"\nTaille de la mantisse : {mp.prec} bits")
print(f"Valeur de réference avec {mp.dps} chiffres significatifs : {mp.nstr(gref, mp.dps, strip_zeros=False)}")
# Evaluation des fonctions
x = mp.mpf('100_000_000')
print(f"1ere forme avec {mp.dps} chiffres significatifs : g(x={x}) = {mp.nstr(g1(x), mp.dps, strip_zeros=False)}")
print(f"2eme forme avec {mp.dps} chiffres significatifs : g(x={x}) = {mp.nstr(g2(x), mp.dps, strip_zeros=False)}")
```

## Evaluation d'une fonction bis

On calcule l'évaluation de la fonction : $f(x,y) = 9 x^4  - y^4 + 2 x^2$ pour différentes précisions :

```python
def f(x,y):
    return 9*x**4 - y**4 + 2*y**2

# flottants sur 32 bits 
mp.prec = 24
x = mp.mpf('40545')
y = mp.mpf('70226')
print(f"Evaluation de la fonction pour des flottants sur  32 bits : f(x,y) = {f(x,y)}")

# flottants sur 64 bits 
mp.prec = 53
x = mp.mpf('40545')
y = mp.mpf('70226')
print(f"Evaluation de la fonction pour des flottants sur  64 bits : f(x,y) = {f(x,y)}")

# flottants sur 128 bits 
mp.prec = 113
x = mp.mpf('40545')
y = mp.mpf('70226')
print(f"Evaluation de la fonction pour des flottants sur 128 bits : f(x,y) = {f(x,y)}")
```

On change l'algorithme de calcul en factorisant le terme $9 x^4  - y^4$

```python
def f(x,y):
    return (3*x**2 - y**2)*(3*x**2 + y**2) + 2*y**2

# flottants sur 32 bits
mp.prec = 24
x = mp.mpf('40545')
y = mp.mpf('70226')
print(f"Evaluation de la fonction pour des flottants sur 32 bits  : f(x,y) = {f(x,y)}")

# flottants sur 64 bits 
mp.prec = 53
x = mp.mpf('40545')
y = mp.mpf('70226')
print(f"Evaluation de la fonction pour des flottants sur 64 bits  : f(x,y) = {f(x,y)}")
```
